## 로컬 변수와 글로벌 변수

````python
EPOCH = 2
BATCH = 5 # 글로벌 변수

def train(weight): # weight는 매개변수, 여기서 weight=0을 실행
    x=1 # x는 로컬 변수
    for i in range(EPOCH): #글로벌 변수는 안에서도 쓸 수 있다.
        for j in range(BATCH):
            print(f"{i}번째 Epoch이며 {j}번째 batch입니다.")
            weight += 1
    return weight
train(0)
print(x) # 로컬 변수라 밖에서 보면 없는 존재
````

````python
def f():
    global x # 안에서 글로벌 변수를 만들 수 있다 (밖에서는 굳이 global을 안 붙여도 알아서 글로벌)
    x = 20
    print(x)
f()
print(x)
````

````python
class Robot:
    def __init__(self, height, sentence): # 밖에서 넘어오는 매개변수는 무조건 self 뒤에
        self.height = height # self.height는 밖에서 들어온 height랑은 다른 거임
        # 클래스 내부에서 써먹을라고 주머니에 넣은 것
        self.sentence = sentence
    def greeting(self):
        print(self.sentence)
        print(f"제 키는 {self.height}이에요.")


bot1 = Robot(180, "안녕")
bot1.greeting()

print(bot1.height)
print(bot1.sentence)
bot1.height -=10
bot1.greeting()

bot2 = Robot(160, "반갑습니다")
bot2.greeting()
````

````python
# 상솟 -> 효율화
class Robot: # 부모 클래스
    def greeting(self):
        print("hello")

class Study_bot(Robot): # 자식 클래스. 로봇 클래스를 상속 받음 -> Robot이 가지는 함수 사용 가능!
    def study(self):
        print("열심열심")

sBot = Study_bot()
sBot.greeting() # Study_bot 클래스에는 분명히 greeting 함수가 없지만 물려받았기 때문에 사용 가능!
sBot.study()
````

````python
# 매소드 오버라이딩(overriding)
class Robot:
    def greeting(self):
        print("hello")

class Study_bot(Robot):
    def greeting(self): # 부모의 greeting을 물려받았으나 여기서 다시 정의하면 덮어써버림
        print("공부하는 로봇입니다.")

bot1 = Study_bot()
bot1.greeting()
bot2 = Robot()
bot2.greeting()
````

````python
# 매소드 오버라이딩(overriding)
class Robot:
    def greeting(self):
        print("hello")

class Study_bot(Robot):
    def greeting(self):
        super().greeting() # super()가 부모 클래스를 가리킴 # 부모의 것을 이렇게 사용 가능
        print("공부하는 로봇입니다.") # 부모의 원래 기능은 유지, 거기에 덧붙이고 싶을 때 오버라이딩 사용 가능

bot1 = Study_bot()
bot1.greeting()
bot2 = Robot()
bot2.greeting()
````

````python
# 딥러닝 코드에선 아래와 같은 코드를 많이 보게 됨
class Robot:
    def __init__(self):
        print("부모 클래스의 init 실행")
        self.money = 100

class Study_bot(Robot):
    def __init__(self):
        super().__init__() # 부모 클래스의 init을 실행
        print("자식 클래스의 init 실행")
        self.book = 10

bot_parent = Robot()
print(bot_parent.money)
bot_child = Study_bot()
print(bot_child.money)
print(bot_child.book)
````

````python
class Robot:
    def __init__(self):
        print("부모 클래스의 init 실행")
        self.money = 100

class Study_bot(Robot):
    def func2(self): # 자식 클래스의 __init__이 없다면 부모 클래스의 __init__ 이 자동 호출
        print("자식 클래스의 init 실행")
        self.book = 10

bot1 = Study_bot()
bot1.func2()
print(bot1.money)
print(bot1.book)
````
